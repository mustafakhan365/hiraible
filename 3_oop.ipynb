Class Methods
-

A class method is a decorator whose first argument is class itself. Its one of the use is that it can access and change the class instance variable.

```python
class man:
    nation="pakistan"

    def __init__(self,a,b):
        self.height=a
        self.weight=b
    def manspecs(self):
        return f"man height is {self.height}. weight is {self.weight}"
    @classmethod
    def change(cls,new):
      cls.nation=new
```

```python
saad=man(4,5)
```

```python
saad.change("USA")
```

```python
saad.nation
```

```python
ali=man(5,6)
```

```python
ali.change("KSA")
```

```python
ali.nation
```

```python
hamza=man(6,7)
```

```python
hamza.nation
```

```python
class man:
    colour = 5  # Class variable (shared by ALL objects)

    def __init__(self, a, b):
        self.height = a
        self.weight = b

    def manspecs(self):
        return f"man height is {self.height}. weight is {self.weight}"

    @classmethod
    def rr(cls, new):
        cls.colour = new  # Changes class variable for ALL objects

ali = man(4, 5)
hamza = man(6, 70)

print(ali.colour)    # 5 (original)
ali.rr(4)            # changing colour via classmethod
print(ali.colour)    # 4
print(hamza.colour)  # 4 (changed for ALL objects!)
```

Class Method as an alternative Constructor
-

```python
class man:
    colour = 5

    def __init__(self, a, b):
        self.height = a
        self.weight = b

    def manspecs(self):
        return f"man height is {self.height}. weight is {self.weight}"

    @classmethod
    def rr(cls, new):
        sp = new.split("-")   # "44-55" will be ["44", "55"]
        return cls(sp[0], sp[1])  # same as man("44", "55")

ali = man.rr("44-55")  # Alternative way to create object
print(ali.height)      # 44
```

Static method:
-

A static method in python is a method which belongs to class only rather than instances of class. You can access it directly with the class name. It can be called by the instances but not recommended.

```python
class man:
    colour = 5

    def __init__(self, height, weight):
        self.height = height
        self.weight = weight

    @staticmethod
    def info():
        print("This is a man class")  # No self, No cls

ali = man(5, 65)

# Access directly with class name
man.info()   # This is a man class

# Also works with object (but not recommended)
ali.info()   # This is a man class
```

Encapsulation and abstraction:
-

Encapsulation and abstraction are two different concepts in object oriented programing but similar to each other. Abstraction is not possible to achieve without encapsulation.

Encapsulation means that storing the data and all of the code related to that data at one place which should be restricted from outer word. But abstraction means the hiding the all internal functionalities and show only essential information.

Encapsulation:
-

```python
class bike():
    def __init__(self, model, cc):
        self.model = model   # data
        self.cc = cc         # data

    def start(self):         # code related to data
        print("bike is starting")

honda = bike(23, 150)
print(honda.model)  # 23
honda.start()       # bike is starting
```

Abstraction:
-

```python
class bike():
    def __init__(self):
        self.name = "honda"

    def specs(self):
        self.cc = "150"
        self.color = "black"
        print(self.name, self.cc, self.color)  # only shows result

    def owner(self):
        self.age = "22"
        self.work = "ML engineer"
        print(self.name, self.age, self.work)  # only shows result

hi = bike()
hi.specs()    # honda 170 black
hi.owner()   # honda 22 ML engineer
```

Inheritance:
-

Inheritance in python means that deriving a new class from the existing class. The existing class is called parent class and the derived class is called child class. By using this concept you can inherit the properties of existing class to new class.

Reduce the programmers efforts.
Increase the code reusability.


image.png

Single Inheritance:
-

Single inheritance means that deriving one new class from the existing class.

```python
class product:
  p = "cake"
  prod2 = "roti"

class property(product):
  q = "sweet"
  prop2 = "healthy"

a=property()
print(a.p + " is " + a.q)
```

```python
class product:          # Parent class
    prod1 = "cake"
    prod2 = "roti"

class property(product):  # Child class
    prop1 = "sweet"
    prop2 = "healthy"

gg = property()
print(gg.prod1 + " is " + gg.prop1)
```

Multiple Inheritance:
-

If a class is derived or inherited from more than one existing class than it is multiple inheritance.

```python
class name:
    first = "ali"
    second = "usman"
    third = "umer"

class grade:
    position1 = "first"
    position2 = "second"
    position3 = "third"

class result(name, grade):   # inheriting from TWO classes
    prize1 = "10000"
    prize2 = "5000"
    prize3 = "3000"

res = result()
print(res.first + "\n" + res.position1 + "\n" + res.prize1)
```

```python
res=result()
print(res.first)
```

Multi-Level Inheritance:
-

Multi-level inheritance means that deriving a new class from existing class which will be called as child class. But here we will derive a grandchild class from the child class.

Deriving a class from another class which is already a child of some other class.

👴 → 👨 → 👦

```python
class name:              # Grandparent
    first = "ali"
    second = "usman"
    third = "umer"

class grade(name):       # Parent (inherits from name)
    position1 = "first"
    position2 = "second"
    position3 = "third"

class result(grade):     # Child (inherits from grade)
    prize1 = "10000"
    prize2 = "5000"
    prize3 = "3000"

res = result()
print(res.first + "\n" + res.position1 + "\n" + res.prize1)
```

Hierarchical Inheritance:
-

If we derive more than one child class from the same parent class then it is called hierarchical inheritance.

One parent, many children — all children share the same parent but have their own unique attributes.

```python
class name:            # Parent
    first = "ali"
    second = "usman"
    third = "umer"

class grade(name):     # Child 1
    position1 = "first"
    position2 = "second"
    position3 = "third"

class result(name):    # Child 2
    prize1 = "10000"
    prize2 = "5000"
    prize3 = "3000"

res = result()
print(res.first + "\n" + res.prize1)
```

Hybrid Inheritance:
-

Hybrid inheritance contains more than one form of inheritance. There is no sequence that which class will be derived from whom. You can use it according to your requirements.

Hybrid is just a mix of any inheritance types used together according to your requirements.

```python
class PC:                        # Grandparent
    pass

class Laptop(PC):                # Parent (Single)
    pass

class Mouse(Laptop):             # Child (Multilevel)
    pass

class Student3(Mouse, Laptop):   # (Multiple)
    pass

obj = Student3()
```

Access Specifiers:
-

Access specifier in python are used to limit the access of class variables and class method outside the class while implementing the concept of inheritance.

Public access specifiers:
All the methods and variables in class are public by default. You can access variables and methods by any child class.


Protected access specifiers:
By using protected access specifiers, you can access the variables and methods inside the class by the class itself and child class but not any other class. It can achieve by putting an underscore after self. Keyword.


Privates access specifiers:
The variables and methods inside the class can only be achieve by class itself but not by any child class or any other class. It can be achieved by putting two underscores after self keyword.


```python
class var:
    def __init__(self):
        self.age="13"
        self._name="ali"
        self.__height=15
class var2(var):
    pass
obj=var2()
print(obj._name)
```

Super() function:
-
If you want to access the parent class variables, methods or constructor from child class having same name, we use super function.


```python
class var:
    x = "hello"          # Parent variable

class var2(var):
    x = "good"           # Child variable (same name)

    def __init__(self):
        print(self.x)        # good  → child variable
        print(super().x)     # hello → parent variable

obj = var2()
```

Simple Rule: When child and parent have same name, use super() to reach the parent's version.

Method Overriding:
-

Child class redefines parent class method with same name.

```python
class Animal:
    def sound(self):
        print("Animal makes a sound")
```

```python
class Dog(Animal):
    def sound(self):             # overrides parent
        print("Dog barks ")
```

```python
class Cat(Animal):
    def sound(self):             # overrides parent
        print("Cat meows ")
```

```python
class Cow(Animal):
    def sound(self):             # overrides parent
        print("Cow moos")
```

```python
animals = [Dog(), Cat(), Cow()]
for i in animals:
    i.sound()
```

```python
animals = [Animal(), Dog(), Cat(), Cow()]

for animal in animals:
    animal.sound()
```

Abstract Class:
-

Abstract class in python is a class which contain one or more abstract method. These abstract methods have an empty body. If you want to run a method forcefully in any child class inherited from the abstract class, you can use this class. You have to import ABC module from abc library

```python
from abc import ABC, abstractmethod
```

```python
from abc import ABC, abstractmethod

class caus(ABC):              # Abstract class
    @abstractmethod
    def mul(self):            # Abstract method (empty body)
        pass

class RR(caus):               # Child class
    def __init__(self):
        self.a = 4
        self.b = 5

    def mul(self):            # Must implement mul()
        return self.a * self.b

obj = RR()
print(obj.mul())
```

Decorators:
-

Decorators are the special type of functions use to enhance the functionality of a function. You can add extra functionality in the function without changing the original function code.

```python
def bold(sd):
    def wrapper():
        print("--- START ---")
        sd()
        print("--- END ---")
    return wrapper

@bold
def greet():
    print("Hello saad!")
```

```python
greet()
```
