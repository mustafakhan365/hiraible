Introduction to NumPy
-

NumPy (Numerical Python) is a library, a collection of ready-made tools written by other people that we can use in our own code.

It gives Python a powerful new data structure called the array, plus fast math functions to work on it.

It's the foundation of almost the entire data science and ML ecosystem. Pandas, scikit-learn, TensorFlow, PyTorch all build on top of NumPy.

```python
import math

# print(math.sqrt(16))
print(math.pi)
```

math is a built-in library that ships with Python. import means "bring this toolbox into my program." We then use its tools with dot notation

```python
import numpy as np

print(np.__version__)
```

NumPy array:

```python
arr = np.array([1, 2, 3, 4, 5])

print(arr)
print(type(arr))

# np.array([0,2],[2,3],[5,6])
```

It looks like a list, but type() shows numpy.ndarray.

Arrays can be 1D (a row of numbers), 2D (a table), or more.

```python
arr = np.array([[10, 20], [30, 40]])

print(arr.shape)
print(arr.dtype)
print(arr.ndim)
```

shape is the size of the array in each dimension.
dtype is the data type of the elements. Unlike lists, a NumPy array stores one data type for all its elements.

```python
my_list = [1, 2, 3]
my_array = np.array([1, 2, 3])

print(my_list * 2)
print(my_array * 2)
```

```python
my_list = [1, "hello", 3.5, True]
print(my_list)

my_array = np.array([1, "hello", 3.5, True])
print(my_array)
print(my_array.dtype)
```

NumPy silently converted everything to strings (<U32 means a text type).

Arrays pick one type that fits all the values.

```python
a = [1, 2, 3]
b = [4, 5, 6]
print(a + b)

x = np.array([1, 2, 3])
y = np.array([4, 5, 6])
print(x + y)
```

NumPy's loops run in compiled C code instead of Python, and the data sits in one continuous block of memory.

```python
import time

size = 5000000
py_list = list(range(size))
np_array = np.arange(size)

start = time.time()
result_list = [x * 2 for x in py_list]
list_time = time.time() - start

start = time.time()
result_array = np_array * 2
array_time = time.time() - start

print(f"List time:  {list_time:.4f} seconds")
print(f"NumPy time: {array_time:.4f} seconds")
print(f"NumPy is about {list_time / array_time:.0f}x faster")
```

element-wise arithmetic:

```python
a = np.array([1, 2, 3, 4])
b = np.array([10, 20, 30, 40])
```

```python
print(a + b)    # [11 22 33 44]
print(b - a)    # [ 9 18 27 36]
print(a * b)    # [ 10  40  90 160]
print(b / a)    # [10. 10. 10. 10.]
print(a ** 2)   # [ 1  4  9 16]
print(b % 3)    # [1 2 0 1]
```

math functions that work on whole arrays:

```python
arr = np.array([1, 4, 9, 16])
```

```python
print(np.sqrt(arr))
print(np.abs(np.array([-3, 2, -1])))
print(np.round(np.array([1.234, 5.678]), 2))
```

aggregation functions:

```python
print(arr.sum())      # 108
print(arr.mean())     # 18.0
print(arr.min())      # 4
print(arr.max())      # 42
print(arr.argmax())   # 5  -> the *index* of the largest value
print(arr.argmin())   # 0  -> the *index* of the smallest value
```

aggregating along an axis (2D data):

```python
matrix = np.array([[1, 2, 3],
                   [4, 5, 6]])
```

```python
print(matrix.sum())          # 21        -> everything
print(matrix.sum(axis=0))    # [5 7 9]   -> down the columns
print(matrix.sum(axis=1))    # [ 6 15]   -> across the rows
print(matrix.mean(axis=0))   # [2.5 3.5 4.5]
```

Reading & Writing Files
-

writing and reading a text file

```python
with open("notes.txt", "w") as f:
    f.write("Line one\n")
    f.write("Line two\n")

with open("notes.txt", "r") as f:
    content = f.read()

print(content)
```

```python
with open("notes.txt","w") as f:
  f.write("Hey how are you\n")
  f.write("How are you?\n")
```

```python
with open("notes.txt","r") as f:
  content = f.read()
print(content)
```

append mode:

```python
with open("notes.txt", "a") as f:
    f.write("\nLine three\n")

with open("notes.txt") as f:
    print(f.read())
```

```python
with open("notes.txt") as f:
    print(f.readline())    # just the first line

with open("notes.txt") as f:
    print(f.readlines())   # list of all lines: ['Line one\n', 'Line two\n', 'Line three\n']

with open("notes.txt") as f:
    for line in f:         # best habit: loop over the file itself
        print(line.strip())
```

CSV with the built-in csv module

```python
import csv

rows = [
    ["id", "name", "age", "city", "marks"],
    [1, "Ali", 20, "Islamabad", 85.5],
    [2, "Sara", 22, "Lahore", 92.0],
    [3, "Ahmed", 21, "Karachi", 78.5],
    [4, "Zara", 23, "Islamabad", 88.0],
]

with open("students.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerows(rows)
```

```python
with open("students.csv", newline="") as f:
    reader = csv.reader(f)
    for row in reader:
        print(row)
```

newline="" is a small detail that stops the csv module from producing blank lines on some systems.

APIs 1, REST Basics & requests
-

An API (Application Programming Interface) is a way for one program to ask another program for data. You send a request, and you get data back.

On the web, most APIs follow the REST style, using normal web addresses (URLs) and returning data, usually as JSON.

Analogy: a restaurant. You (the program) don't walk into the kitchen. You give an order to the waiter (the API), and the waiter brings back the food (the data).

https://jsonplaceholder.typicode.com/posts?userId=1

Base URL: the address of the server

Endpoint: the specific resource you want (/posts, /users)

Parameters: filters after the ?, written as key=value

HTTP methods:

our first API call:

```python
import requests

response = requests.get("https://jsonplaceholder.typicode.com/posts/15")
print(response)
print(response.status_code)
print(response.json())
```

```python
print(response)
```

requests is a library that makes web requests simple.

status_code tells us whether it worked. 200 means success.

.json() converts the JSON text into a normal Python dictionary.

parsing the response:

```python
post = response.json()

print(type(post))
print(post["title"])
print(post["userId"])
```

query parameters:

```python
params = {"userId": 1}
response = requests.get("https://jsonplaceholder.typicode.com/posts", params=params)

posts = response.json()
print(len(posts))
print(response.url)
```

nested JSON:

```python
response = requests.get("https://jsonplaceholder.typicode.com/users/1")
user = response.json()

print(user["name"])
print(user["address"]["city"])
print(user["address"]["geo"]["lat"])
print(user["company"]["name"])
```

Handling errors:

```python
try:
    response = requests.get("https://jsonplaceholder.typicode.com/posts/99999", timeout=5)
    response.raise_for_status()   # raises an error if status is 4xx or 5xx
    print(response.json())
except requests.exceptions.HTTPError as e:
    print("HTTP error:", e)
except requests.exceptions.RequestException as e:
    print("Request failed:", e)
```

POST Request:

```python
new_post = {"title": "My first post", "body": "Hello API", "userId": 1}

response = requests.post("https://jsonplaceholder.typicode.com/posts", json=new_post)

print(response.json())
```
