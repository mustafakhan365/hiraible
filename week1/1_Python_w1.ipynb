Conditionals
-

Conditionals let the program make decisions run certain code only if something is true.

```python
age = 20

if age >= 18:
    print("You are an adult.")
```

```python
marks = 75

if marks >= 90:
    print("Grade: A")
elif marks >= 75:
    print("Grade: B")
elif marks >= 50:
    print("Grade: C")
else:
    print("Grade: F")
```

nested conditions:
-

```python
age = 25
has_id = True

if age >= 18:
    if has_id:
        print("Entry allowed.")
    else:
        print("Entry denied. Show your ID.")
else:
    print("Entry denied. Too young.")
```

Lists & Loops
-

So far every variable held one value. A list lets us store many values together, in order.

Loops let us repeat code instead of writing the same line over and over.

creating and accessing lists:
-

```python
fruits = ["apple", "banana", "cherry", "mango"]

print(fruits)
print(fruits[0])
print(fruits[-1])
print(len(fruits))
```

Lists use the same indexing rules as strings — starts at 0, negative counts from the end.

modifying lists:
-

```python
fruits = ["apple", "banana", "cherry"]

fruits.append("mango")
print(fruits)

fruits.remove("banana")
print(fruits)

fruits[0] = "orange"
print(fruits)
```

for loops:
-

```python
fruits = ["apple", "banana", "cherry", "mango"]

for fruit in fruits:
    print(fruit)
```

for loop with range():
-

```python
for i in range(5):
    print(i)
```

```python
for i in range(1, 6):
    print(i)
```

while loops:
-

```python
count = 1

while count <= 5:
    print(count)
    count = count + 1
```

combining loop + if:
-

```python
numbers = [10, 15, 22, 33, 40, 7]

for num in numbers:
    if num % 2 == 0:
        print(f"{num} is even")
    else:
        print(f"{num} is odd")
```

Functions:
-

A function is a reusable block of code that we define once and can call as many times as we want.

```python
def greet():
    print("Hello, welcome to the bootcamp!")

greet()
greet()
```

parsing argument:
-

```python
def greet(name):
    print(f"Hello, {name}!")

greet("Ali")
greet("Sara")
```

```python
def add_numbers(a, b):
    print(a + b)

add_numbers(5, 3)
add_numbers(10, 20)
```

```python
# Default parameter

def greet(name="friend"):
    print(f"Hello, {name}!")

greet()
greet("Ali")
```
